```python
# observation from science
# 25% - 25% students scored below 25.25 marks, weak group (0-30)
# 50% - 50% students scored below 53 marks, below average (30-55)
# 75% - 75% students scored below 78 marks, average to good

# helps to understand distribution & performace of groups

# mean == median : balanced / symmetric
# mean >> meadian : right skew, one large value (eg:200) is pulling mean up
# mean << median : left skew, extreme low increases, if some students score extremely high or low

# eg:
#     mean = 51.44
#     median = 53
#     no extreme distortion so data is balanced
    
# std : standard deviation, measures how spread out values (marks) are from mean

# 45,48,50,52,55 : std is small, students perform similarly
# 10,30,50,70,100 : std is high, some are very good, some are very weak
```
