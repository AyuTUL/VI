```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("marksheet.csv")
df.head()
```

```python
plt.figure(figsize=(8, 4))
plt.scatter(df['English'], df['Maths'])
plt.title('Science vs Math')
plt.xlabel('Science')
plt.ylabel('Math')
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.bar(df["Section"],df["Maths"])

plt.xlabel("Sections")
plt.ylabel("Marks in Maths")
plt.title("Student marks comparison across Sections A, B & C")
```

```python
df30=df.head(30)

plt.plot(df30["Maths"],df30["Section"]=="A",marker='o',label="A",linestyle="--")
plt.plot(df30["Maths"],df30["Section"]=="B",marker='o',label="B",linestyle="--")
plt.plot(df30["Maths"],df30["Section"]=="C",marker='o',label="C",linestyle="--")

plt.xlabel("Sections")
plt.ylabel("Marks in Maths")
plt.title("Maths marks comparison : A vs B vs C")

plt.legend()
plt.grid()
plt.show()
```
