```python
# if else
# if <condition>
    # <operation>
# elif <condition>
    # <operation>
# else
    # <operation>
age=int(input("Enter your age : "))
if age>18:
    print("You can vote")
elif age==18:
    print("You can vote after 1 year")
else:
    print("You cannot vote")
```
