```python
# base class: libaryItems
# derived class: book (returnDate)

# libraryItems : title, author, category, price
# methods : displayInfo, applyDiscount, finalPrice

class LibraryItems:
    def __init__(self,t,a,c,p):
        self.title=t
        self.author=a
        self.category=c
        self.price=p
    
    def displayInfo(self):
        print(f"{self.title}\t{self.author}\t{self.category}\t{self.price}")

    def applyDiscount(self,discountRate):
        return discountRate/100*self.price

    def finalPrice(self,discount):
        return self.price-discount

class Book(LibraryItems):
    def __init__(self, t, a, c, p,d):
        super().__init__(t, a, c, p)
        self.returnDate=d

b=Book("Technical Writing for Success","Darlene Smith","Horror, Romance",8000.9009,"6th Aug, 2025")

discount=b.applyDiscount(20)
finalPrice=b.finalPrice(discount)

print(f"Title\t\t\t\tAuthor\t\tCategory\tPrice")
b.displayInfo()

    
```
