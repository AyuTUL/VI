```python
# Base Class : Product
# Child Class :
#     Electronics -> warranty
#     Clothing -> size
# All products have :
#     name
#     price
class Product:
    def __init__(self,name,price):
        self.name=name
        self.price=price
        
    def showProduct(self):
        print(f"Product\tPrice\n------------------")

class Electronics(Product):
    def __init__(self,name,price,warranty):
        super().__init__(name,price)
        self.warranty=warranty
    
    def showElectronics(self):
        print(f"{self.name}\t{self.price}\t{self.warranty}")
        
class Clothing(Product):
    def __init__(self,name,price,size):
        super().__init__(name,price)
        self.size=size
    
    def showClothing(self):
        print(f"{self.name}\t{self.price}\t{self.size}")
        
phone=Electronics("Samsung",50505.11,3)
pants=Clothing("Jeans",2000.121,"XL")

phone.showProduct()
phone.showElectronics()
pants.showClothing()
```
