Polymorphism

```python
class Notification:
    def send(self):
        print("Sending notification")

class Email(Notification):
    def send(self):
        print("Sending email")
        
class SMS(Notification):
    def send(self):
        print("Sending sms")
        
class Push(Notification):
    def send(self):
        print("Push notification")
        
notifications=[Email(),SMS(),Push()]

for n in notifications:
    n.send()
```

```python
# Shape system:
#     Base : shape
#     Child : Circle, Rectangle
#     Method: area() behaves diferently

class Shape:
    def area(shape):
        pass

class Circle(Shape):
    def __init__(self,r):
        self.radius=r
        
    def area(self):
        return 3.14*self.radius*self.radius
    
class Rectangle(Shape):
    def __init__(self,l,b):
        self.length=l
        self.breadth=b
    
    def area(self):
        return self.length*self.breadth
    
shapes=[Circle(7),Rectangle(4,5)]

for s in shapes:
    print("Area = ",s.area())
```
