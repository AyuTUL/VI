## Class
- blueprint used to create objects with similar attributes, methods and behaviour


Example: 
- Class : college, car, student
- Objects : college_name, (toyota, bmw), (ram, sita)

Advantages:
- reusability
- keeps data organized
- add more features
- create unlimited objects

````python
class Student:
    def  __init__(self,name,age,marks):
        self.name=name
        self.age=age
        self.marks=marks
    
    def show(self):
        print(self.name,"\t",self.age,"\t",self.marks)
        
s1=Student("Kimmich",28,50.12)
s2=Student("Vinicius",25,99.22)
s1.show()
s2.show()
````

### __init__ Constructor
-constructor that runs automatically when object is created

When to not use init? 
1. No initial data needed
    ```
    class Hello:
        def greet(self):
            print("hi")
    ```

2. Temporary results

3. Static value
    ```
    def greet(self):
        print("hello")
    ```

````python
class Student:
    def __init__(self,name,age,phone):
        self.name=name
        self.age=age
        self.phone=phone
    
    def display(self):
        print(self.name,"\t",self.age,"\t",self.phone)

name=input("Enter student name : ")
age=int(input("Enter age : "))
phone=input("Enter phone number : ")

s=Student(name,age,phone)

print("\nStudent details : ")

s.display()        
````

````python
# without __init__
class Student:
    def details(self,name,age,marks):
        self.name=name
        self.age=age
        self.marks=marks
    
    def show(self):
        print(f"Name : {self.name}, Age : {self.age}")
        
s1=Student()
s1.details("Kimmich",28,50.12)
s2=Student()
s2.details("Vinicius",25,99.22)
s1.show()
s2.show()
````

### __str__ Method

````python
class Student:
    def __init__(self,name,age,gender):
        self.name=name
        self.age=age
        self.gender=gender
    
    def __str__(self):
        return f"Name : {self.name}, Age : {self.age}, Gender : {self.gender}"
    
a=Student("Benzema",15,"M")
print(a) 
````
