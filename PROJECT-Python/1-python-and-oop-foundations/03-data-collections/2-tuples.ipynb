## Tuple
- immutable
- defined by small brackets

```python
a=(1,2,3,6)
print(a)
print(type(a))
print(a[3])
```

```python
a=(1)
print(type(a))
b=(1,)
print(type(b))
print(len(b))
```

```python
a="hello"
tuple(a)
```

```python
# tuple into list
a=("hello",)
list(a)
```

### List Inside Tuple

```python
a=([3,2],[3,5],[33])
a
```

### Nested Tuple

```python
a=(21,(21,))
a
```
