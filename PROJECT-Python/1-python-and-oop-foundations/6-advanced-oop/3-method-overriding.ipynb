# Method Overriding
- parent has method, child inherits it, child changes same method

```python
class Animal:
    def sound(self):
        print("Animal makes sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

a=Animal()
b=Dog()

a.sound()
b.sound()
```

```python
# using super() with overriding
class Person:
    def show(self):
        print("I am person")

class Student(Person):
    def show(self):
        Person().show()
        print("I am student")

s=Student()
s.show()
```
