```python
# ATM system:
#     parent(account)
#     child(Saving account)
    
#     Data types:
#         public : account holder name
#         protected : _balance
#         private : __pin

# Methods:
#     deposit, withdraw, showBalance

class Account:
    def __init__(self,name,balance,pin):
        self.name=name              # public
        self._balance=balance       # protected
        self.__pin=pin              # private
        
    def deposit(self,amount):
        self._balance+=amount
        print("\nDeposited Amount = Rs.",amount)
        print("Current Amount = Rs.",self._balance)
        
    def withdraw(self,amount):
        if(self._balance<amount):
            print("\nWithdraw unsuccesful : Not enough balance")
        else:
            self._balance-=amount
            print("\nWithdrawn Amount = Rs.",amount)
            print("Current Amount = Rs.",self._balance)
            
    def showBalance(self):
        print("Balance = Rs.",self._balance)
        
class SavingAccount(Account):
    def __init__(self,name,balance,pin):
        super().__init__(name,balance,pin)
        
    def show(self):
        print(self.__pin)

a=SavingAccount("Ayush",9000.90,4275)
a.showBalance()
a.deposit(1000.10)
a.withdraw(5)
a.withdraw(500000)
a.show()                              # wont work since __pin isnt inherited
```
