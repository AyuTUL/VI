# Abstraction
Hide internal details and show only essential details, see only what to use

### Advantages:
- reduce complexity
- improve security
- hide sensitive logic
- make system easier to use

```python
# Hospital System:
#     Base : treatment (abstract)
#     Child : surgery, therapy
    
#     each must implement : performTreatment()

from abc import ABC, abstractmethod

class Treatment(ABC):
    @abstractmethod
    
    def performTreatment(self):
        pass

class Surgery(Treatment):
    def performTreatment(self):
        print("Performing surgery with doctors")

class Therapy(Treatment):
    pass
    # def performTreatment(self):
    #     print("Performing therapy with physios")
        
patient1=Surgery()
patient2=Therapy()

patient1.performTreatment()
patient2.performTreatment()
    
```

#### Decorator:
- modifies / adds behaviour to function without changing function's code

```python
def greet():
    print("hello")
greet()

def myDecorator(func):
    def wrapper():
        print("Before")
        func()
        print("After")
        return wrapper

@myDecorator
def greet():
    print("hello")
greet()
```
