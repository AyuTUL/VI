Function

```python
def display(): # parameter
    print("Hellow")
display() # argument
# parameter is variable in function definition, placeholder
# argument is actual value passed when calling function 
```

```python
def sum(a,b):
    return a+b
print(sum(3,5))
```

Types of Functions

```python
# 1. No parameter, no return
# 2. parameter, no return
# 3. No parameter, return
# 4. parameter, return

def hello1():
    print("No parameter, no return")
    
place1="parameter, no return"
def hello2(place):
    print(place)
    
def hello3():
    return "No parameter, return"

place2="parameter, return"
def hello4(place):
    return place

hello1()
hello2(place1)
print(hello3())
print(hello4(place2))
```

```python
# default parameter function
def greet(name="Student"):
    print("hello",name)
greet()
greet("Sonny")
```

```python
# keyword arguments
def info(name,age):
    print(name,age)
info(name="bale",age=20)
```

```python
# lambda function (short function)
add=lambda a,b:a+b
print("Sum = ",add(2,3))
```

```python
# variable length arguments
# args(multiple values)
def add(*numbers):
    print(sum(numbers))
add(1,-1,0)
```

```python
# kwargs (key-value pair)
def info(**data):
    print(data)
info(name="Sam",age=15)
```
