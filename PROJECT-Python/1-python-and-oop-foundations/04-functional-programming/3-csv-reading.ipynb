```python
import pandas as pd
df=pd.read_csv("pokemon.csv")
print(df)
```

```python
# df.loc[row,column]
# df.iloc[row_position,column_position]

print(df.loc[0,"Name"])
# 0 : index label
# Name : column name

print(df.iloc[1,0])
# 1 : 2nd row position
# 0 : 1st column position
```
