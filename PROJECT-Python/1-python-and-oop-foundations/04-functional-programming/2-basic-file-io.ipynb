## File Handling
```
file=open(<filename>,<mode>)
file.close()

with open(<filename>,<mode>) as file:
    <operations>

modes:
    read->r
    create->x
    write->w
    append->a
```

````python
try:
    file=open("data.txt","x")
    file.close()
except:
    print("File already created")
````

````python
file=open("data.txt","r")
data=file.read()
print(data)
file.close()
````

````python
file=open("data.txt","w")
file.write("python,java,c++,c#")
file.close()
````

````python
try:
    file=open("info.csv","x")
    file.close()
except:
    print("File already exists")
````
