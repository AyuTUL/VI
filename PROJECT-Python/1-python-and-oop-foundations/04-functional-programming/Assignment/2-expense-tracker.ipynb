```python
# expense tracker using function
expense=dict()
total=0.0
expense_number=int(input("Enter the no. number of expenses : "))

for i in range(expense_number):
    expense_name=input("Enter expense name : ")
    expense_amount=float(input("Enter expense amount : "))
    expense[expense_name]=expense_amount
    total=total+expense_amount

print("Expenses\t Amount")
for name,amount in expense.items():
    print(name,"\t\t",amount)
print("----------------------\nTotal\t\t",total)
```
