## Set Operations:
#### 1. Append

```python
s=set()
s.add("C")
s.add("Jav")
s
```

#### 2. Remove

```python
a.remove("Jing")
a
```

#### 3. Update

```python
a={1,2,4}
b={"String","Jing"}
a.update(b)
a
```

#### 4. Intersection, Union, Difference

```python
s={"Bale","Benzema","Cristiano","Messi","Neymar","Suarez"}
t={"Messi","Suarez","Ramos"}
print(s.intersection(t))
print(t.union(s))
print(s.difference(t))
print(s-t)
print(t-s)
```
