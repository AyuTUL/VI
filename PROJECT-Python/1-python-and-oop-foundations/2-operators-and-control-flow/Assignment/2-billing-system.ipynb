### Billing System
- name
- price
- quantity
- total

```python

storage=["apple","mango","kebab"]
price=[100,250,400]

print("Food items :")
print("SN\t Name\t Price\t")
for i in range(len(storage)):
    print(i+1,"\t",storage[i],"\t Rs.",price[i])

choice='y'
total=0
while (choice=='y'):
    item=int(input("Enter item number : "))
    qty=int(input("Enter quantity : "))
    total=total+price[item-1]*qty
    choice=input("Add another item? [y/n] : ")
print("Total = Rs.",total)
```
