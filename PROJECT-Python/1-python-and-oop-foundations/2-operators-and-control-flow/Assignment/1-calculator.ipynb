### Calculator

```python
n1=int(input("Enter 1st number : "))
n2=int(input("Enter 2nd number : "))
operators=["+","-","*","/"]
choice="y"
while (choice=="y"):
    op=input("Enter operator : ")
    if op not in operators:
        print("Invalid operator")
    else:
        if op=="+":
            result=n1+n2
        elif op=="-":
            result=n1-n2
        elif op=="*":
            result=n1*n2
        elif op=="/":
            if n2==0:
                print("Divisor can't be -ve")
                break
            else:
                result=n1/n2
        print("Result :\n",n1,op,n2,"=",result)
    choice=input("Another operation? [y/n] : ")
```
