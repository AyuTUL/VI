# If-Else Statement
```
if <condition>:
    <operation>
elif <condition>:
    <operation>
else:
    <operation>
```

````python
age=int(input("Enter your age : "))
if age>18:
    print("You can vote")
elif age==18:
    print("You can vote after 1 year")
else:
    print("You cannot vote")
````

````python
a=int(input("Enter integer : "))
if a%2==0:
    print(a," is even")
else:
    print(a," is odd")
````
