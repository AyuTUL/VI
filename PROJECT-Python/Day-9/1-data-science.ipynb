Bar Graph

```python
import matplotlib.pyplot as plt
subjects=["Math","Science","English","Computer"]
marks=[85,78,92,88]

plt.figure(figsize=(10,5))
plt.bar(subjects,marks)

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Student marks comparison")
```

```python
# check if values are evenly distributed
# if all bars are similar or tends to be -> balanced
# if one is very high or low -> outliers
```

```python
# when not to use bar graph:
#     data is continuous over time -> use line chart
#     for distribution -> use histogram
#     want relationship -> use scatter plot
```

Line Graph

```python
exams=[1,2,3,4,5]
marks=[60,65,70,68,75]

plt.figure(figsize=(5,3))
plt.plot(exams,marks,marker='^') # s : square, o : circle, ^ : triangle

plt.xlabel("Exam number")
plt.ylabel("Marks")
plt.title("Marks trend over Exams")

plt.show() # not compulsory
```

```python
# used to show how something changes over time or over continuous range or count
# connects data points with lines, making patterns & trends easy to see
# for showing trend (change over time)

# what we can see from line graph:
#     trend (increase / decrease)
#     fluctuations
#     pattern over time
#     rate of change : how fast values increase / decrease
#     pattern / cycle
#     repeated behaviour : (eg : temperature rising / falling every afternoon)
#     best for time series data
#     helps to detect anomalies (sudden spikes / drops)
```

```python
days=[1,2,3,4,5]

apple=[150,155,160,158,165]
google=[140,142,145,150,155]

plt.plot(days,apple,marker='o',label="Apple",linestyle="--")
plt.plot(days,google,marker='s',label="Google")

plt.xlabel("Days")
plt.ylabel("Stock Price")
plt.title("Stock comparison : Apple vs Google")

plt.legend() # shows names on graph
plt.grid()
plt.show()
```

```python
# trend comparison:
#     apple rising faster than google
#     if one line is always above : better performance
    
# crossover : when 2 lines intersect
    
# volatility risk:
#     how much line fluctuates
#     more zigzag : more unstable
#     smooth : stable
    
# divergence / convergence:
#     lines moving apart : divergence
#     lines coming together : convergence
```

Scatter Plot

```python
volume1=[100,200,300,400,500,600,700]
price_change1=[1,2,1.5,3,2.5,3.5,4]

volume2=[100,200,300,400,500,600,700]
price_change2=[0.5,1,1.2,1.8,2,2.2,2.5]

z=[260,490,550,450,530,610,730]

plt.figure(figsize=(5,5))
plt.scatter(volume1,price_change1,label="Stock A")
plt.scatter(volume2,price_change2,label="Stock B")

plt.xlabel("Trading Volume")
plt.ylabel("Price Change")
plt.title("Scatter Plot : Volume vs Price Change")

plt.legend()
plt.grid()
plt.show()
```

```python
# scatter plot shows relationship between 2 variables
# each point : one observation
# x axis : one feature
# y axis : another feature

# shows how things are related
```

```python
# upward patterns : positive correlation
# downward pattern : negative correlation
# random dots : no relation
# far point : outliers
# groups : clusters

# not about time but about relationship between variables 
```
