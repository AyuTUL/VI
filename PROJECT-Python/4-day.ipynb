```python
# set operations
s={"Bale","Benzema","Cristiano","Messi","Neymar","Suarez"}
t={"Messi","Suarez","Ramos"}
print(s.intersection(t))
print(t.union(s))
print(s.difference(t))
print(s-t)
print(t-s)
```

```python
# create a dictionary to store student multiple scores
table=dict()
n=int(input("Enter no. of students : "))
for i in range(n):
    name=input("Enter name : ")
    score1=input("Enter score1 : ")
    score2=input("Enter score2 : ")
    score3=input("Enter score3 : ")
    table[name]=[score1,score2,score3]
print("\nStudent Details :\n")
print("Name\t C++\t DSA\t Maths")
for student,subject in table.items():
    print(student,"\t",subject[0],"\t",subject[1],"\t",subject[2],"\t")
    
```

```python
# type of function
# 1. No parameter, no return
# 2. parameter, no return
# 3. No parameter, return
# 4. parameter, return

def hello1():
    print("No parameter, no return")
    
place1="parameter, no return"
def hello2(place):
    print(place)
    
def hello3():
    return "No parameter, return"

place2="parameter, return"
def hello4(place):
    return place

hello1()
hello2(place1)
print(hello3())
print(hello4(place2))
```
