```python
# bank account system
# create program to manage bank account
# attributes:
#     account holder Name
#     type of account
#     balance
# program should:
#     create multiple accounts
#     deposit money
#     withdraw money
#     display account details  
class Account:
    def __init__(self,holder_name,type,balance):
        self.holder_name=holder_name
        self.type=type
        self.balance=balance
    
    def deposit(self,amount):
        self.balance=self.balance+amount
        print(f"Rs. {amount} has been successfully deposited to {self.holder_name}'s account")
    
    def withdraw(self,amount):
        if self.balance >= amount:
            self.balance=self.balance-amount
            print(f"Rs. {amount} has been successfully withdrawn from {self.holder_name}'s account")
        else:
            print(f"Withdraw unsuccessful : Balance is only {self.balance}")
        
    def display(self):
        print(f"Account Details :\nAccount Holder : {self.holder_name}\nType of Account : {self.type}\nBalance : Rs. {self.balance}")
        
table=list()

choice="y"
while True:
    print("\n---Bank Account Management System---\n1. Create Account\n2. Deposit\n3. Withdraw\n4. View Account Details\n5. Exit")
    menu_choice=int(input("Enter your choice : "))

    if menu_choice==1:
        print("\n---Account Creation---\nEnter account details : ")
        name=input("Enter account holder's name : ")
        type=input("Enter type of account : ")
        balance=float(input("Enter balance : "))

        a=Account(name,type,balance)
        table.append(a)

    elif menu_choice==2:
        print("\n---Deposit Operation---")
        name=input("\nEnter account holder's name : ")
        amount=float(input("Enter deposit amount : "))
        for account in table:
            if account.holder_name==name:
                found=True
                account.deposit(amount)
            else:
                found=False
        if not found:
            print("Deposit unsuccessful : Account not found")

    elif menu_choice==3:
        print("\n---Withdraw Operation---")
        name=input("Enter account holder's name : ")
        amount=float(input("Enter withdraw amount : "))
        for account in table:
            if account.holder_name==name:
                found=True
                account.withdraw(amount)
            else:
                found=False
        if not found:
                print("Withdraw unsuccessful : Account not found")

    elif menu_choice==4:
        print("\n---View Account Details---")
        name=input("Enter account holder's name : ")
        for account in table:
            if account.holder_name==name:
                found=True
                account.display()
            else:
                found=False
        if not found:
            print("Display unsuccessful : Account not found")

    elif menu_choice==5:
        print("Thank you for using our bank")
        break

    else:
        print("Invalid choice. Please try again [1-4]")
```
