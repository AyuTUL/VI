```python
# inheritance
# one class and reus and extend another class
# child class gets properties and behaviour from parent class

# Example:
#     parent (base class):
#         Person:
#             name
#             age

#     child clases:
#         Student
#         Teacher


# both Student and Teacher have name and age through Person but can also have extra features
# code reusability
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_person(self):
        print(f"Name : {self.name}, Age : {self.age}")


class Student(Person): # inheriting person
    def __init__(self,name,age,roll):
        super().__init__(name,age) # calling parent constructor
        self.roll=roll
    
    def display_student(self):
        print(f"Roll : {self.roll}")

s1=Student("Endrick",19,2011)
s1.display_person()
s1.display_student()
```
