# Bar Graph
Check if values are evenly distributed:

- if all bars are similar or tends to be -> balanced
- if one is very high or low -> outliers

#### When not to use bar graph:
- data is continuous over time -> use line chart
- for distribution -> use histogram
- want relationship -> use scatter plot

```python
import matplotlib.pyplot as plt
```

```python
subjects=["Math","Science","English","Computer"]
marks=[85,78,92,88]

plt.figure(figsize=(10,5))
plt.bar(subjects,marks)

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Student marks comparison")
```

# Line Graph
- used to show how something changes over time or over continuous range or count
- connects data points with lines, making patterns & trends easy to see
- for showing trend (change over time)

#### What we can see from line graph:
- trend (increase / decrease)
- fluctuations
- pattern over time
- rate of change : how fast values increase / decrease
- pattern / cycle
- repeated behaviour : (eg : temperature rising / falling every afternoon)
- best for time series data
- helps to detect anomalies (sudden spikes / drops)

```python
exams=[1,2,3,4,5]
marks=[60,65,70,68,75]

plt.figure(figsize=(5,3))
plt.plot(exams,marks,marker='^') # s : square, o : circle, ^ : triangle

plt.xlabel("Exam number")
plt.ylabel("Marks")
plt.title("Marks trend over Exams")

plt.show() # not compulsory
```

```python
days=[1,2,3,4,5]

apple=[150,155,160,158,165]
google=[140,142,145,150,155]

plt.plot(days,apple,marker='o',label="Apple",linestyle="--")
plt.plot(days,google,marker='s',label="Google")

plt.xlabel("Days")
plt.ylabel("Stock Price")
plt.title("Stock comparison : Apple vs Google")

plt.legend() # shows names on graph
plt.grid()
plt.show()
```
