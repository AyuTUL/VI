# Heatmap

uses colors to represent numerical values in table / matrix

used when values are high, low or showing patterns across 2 dimensions

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    "maths" : [49,40,96,33,21],
    "science" : [21,29,12,15,43],
    "english" : [81,41,87,53,6],
    "history" : [62,17,16,82,3]
}

df=pd.DataFrame(data)

correlation=df.corr()
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Heatmap of Student Marks")
```

## Correlation

correlation measures linear relationship

- correlation = +1 , perfect positive relationship
- correlation = 0 , no linear relationship
- correlation = -1 , perfect negative relationship

0.3 : weak positive relationship

### Scatter : study relationship between 2 variables
### Heatmap : compare relationship among many variables
