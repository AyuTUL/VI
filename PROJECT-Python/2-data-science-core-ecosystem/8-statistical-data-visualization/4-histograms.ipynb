# Histogram

```python
import matplotlib.pyplot as plt
```

```python
math_marks=[12,18,21,25,27,29,31,34,35,37,
            39,80,12,56,23,78,45,23,89,56,
            90,56,34,89,88,45,68,23,12,18]

plt.hist(
    math_marks,
    bins=[0,20,40,60,80,100],
    edgecolor="black"
)

plt.xlabel("Maths Marks")
plt.ylabel("No. of Marks")
plt.title("Histogram of Marks in Maths")
plt.xticks([0,20,40,60,80,100])
plt.show()
```
