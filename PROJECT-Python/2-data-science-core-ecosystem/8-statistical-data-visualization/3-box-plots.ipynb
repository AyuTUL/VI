# Box Plot
- q1 = 25th percentile
- q2 = 50th percentile
- q3 = 75th percentile

#### Box size : how spread out the middle 50% is
- Small box : median values are close together
- Larger box : variation in median values
    
#### Whisker : lines extending from box, shows range of data within the non-outlier's limits

#### Outliers : values that are unusually far from the rest

Eg : if median=57.4, 50% of students scored below 57.4 and 50% scored above

```python
import matplotlib.pyplot as plt
```

```python
science=[40,45,50,55,60,65,70,75,90,150]
math=[42,47,52,58,62,67,72,75]

plt.figure(figsize=(6,4))
plt.boxplot(science)
plt.title("Box Plot")
plt.show()
```
