# Correlation

- measures linear relationship
- not about time but about relationship between variables

- correlation = +1 , perfect positive relationship
- correlation = 0 , no linear relationship
- correlation = -1 , perfect negative relationship

0.3 : weak positive relationship

#### Using Scatter Plot:
- Upward patterns -> positive correlation
- Downward pattern -> negative correlation
- Random dots -> no relation
- Far points -> outliers
- Group -> clusters 


## Positive Correlation
- as x increases -> y increases

eg : as temperature increases, ice-cream sales increases

```python
import matplotlib.pyplot as plt
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6,7]
y=[2,4,5,4,6,7,8]

plt.scatter(x,y,color="green")
plt.title("positive correlation")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

## Negative Correlation
Eg : as speed increases, time decreases

```python
x=[1,2,3,4,5,6,7]
y=[10,8,6,4,5,2,1]

plt.scatter(x,y,color="green")
plt.title(" negative correlation")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

## No Correlation
- no pattern at all (random)
- one cannot predict the other
Eg : shoe size vs IQ

```python
import random
x=[1,2,3,4,5,6,7]
y=[random.randint(1,10) for i in x]

plt.scatter(x,y,color="green")
plt.title(" no correlation (random)")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```
