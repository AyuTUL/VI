# Encoder
- converts string into numerical form

## Types of Encoding:
- Label Encoding : one column becomes numbers
- One-hot Encoding : creates separate columns

### Label Encoding

```python
import pandas as pd
df=pd.read_csv("../../Datasets/marksheet.csv")
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()

df['Gender'] = le.fit_transform(df['Gender'])
df['Section'] = le.fit_transform(df['Section'])
df
```

### One-hot Encoding

```python
df=pd.get_dummies(df,columns=["Gender","Section"],dtype=int)
df
```
