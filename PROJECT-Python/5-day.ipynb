```python
# default parameter function
def greet(name="Student"):
    print("hello",name)
greet()
greet("Sonny")
```

```python
# keyword arguments
def info(name,age):
    print(name,age)
info(name="bale",age=20)
```

```python
# lambda function (short function)
add=lambda a,b:a+b
print("Sum = ",add(2,3))
```

```python
# variable length arguments
# args(multiple values)
def add(*numbers):
    print(sum(numbers))
add(1,-1,0)
```

```python
# kwargs (key-value pair)
def info(**data):
    print(data)
info(name="Sam",age=15)
```

```python
# student result system:
# create a function that:
#     takes marks as input
#     returns "Pass" or "Fail"
subject=dict()

name=input("Enter student name : ")
subject_number=int(input("Enter no. of subjects : "))

for i in range(subject_number):
    
    subject_name=input("Enter subject name : ")
    mark=float(input("Enter marks : "))
    subject[subject_name]=mark

def grade(subject_number,**subject):
    
    percentage=sum(subject.values())/subject_number
    if percentage>=40.00:
        return "Pass"
    else:
        return "Fail"

print(grade(subject_number,**subject))
```

```python
# login system (username check)
# System checks if username is correct
# write function that verifies username
username_set={"ayush","byush","cyush","dyush"}
username=input("Enter username : ")

def login(username):
    if username in username_set:
        print("User found : Login succesful")
    else:
        print("User not found : Login failed")
        
login(username)
```
