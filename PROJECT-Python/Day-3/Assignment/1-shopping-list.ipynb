```python
#create a python program that:
# 1) allows the user to enter 5 shopping item
# 2) for each item, ask for :
# item name
# price
# quantity
# 3)store the data in lists
# 4) display the complete shopping with price and quantity
# 5) calculate the total cost for each item
# 6) calculate the final amount
# 7) ask the user to search for an item and use if-else to check if it exits.
item=list()
final=0.0

n=int(input("Enter no. of items : "))

for i in range(n):

    name=input("Enter item name : ")
    price=float(input("Enter price : "))
    quantity=int(input("Enter quantity : "))

    total=price*quantity;
    final=final+total

    item.append([name, price, quantity,total])

print("Item\t Price\t Quantity\t Total")

for i in range(n):
    print(item[i][0],"\t",item[i][1],"\t",item[i][2],"\t\t",item[i][3])
print("\nFinal Total =", final)

choice = input("Search item? [y/n] : ")
while choice.lower() == 'y':
    key = input("Enter item name : ")
    found = False

    for item_name in item:
        if key.lower() == item_name[0].lower():
            found = True
            print("\nItem found :")
            print("Name : ", item_name[0])
            print("Price : ", item_name[1])
            print("Quantity : ", item_name[2])
            print("Total : ", item_name[3])
            break

    if not found:
        print("\n",key," not found")

    choice = input("Search another item? [y/n] : ")
```
