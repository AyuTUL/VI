```python
# calculator
n1=int(input("Enter 1st number : "))
n2=int(input("Enter 2nd number : "))
operators=["+","-","*","/"]
choice="y"
while (choice=="y"):
    op=input("Enter operator : ")
    if op not in operators:
        print("Invalid operator")
    else:
        if op=="+":
            result=n1+n2
        elif op=="-":
            result=n1-n2
        elif op=="*":
            result=n1*n2
        elif op=="/":
            if n2==0:
                print("Divisor can't be -ve")
                break
            else:
                result=n1/n2
        print("Result :\n",n1,op,n2,"=",result)
    choice=input("Another operation? [y/n] : ")
```

```python
# billing system (name,price,quantity,total)
storage=["apple","mango","kebab"]
price=[100,250,400]

print("Food items :")
print("SN\t Name\t Price\t")
for i in range(len(storage)):
    print(i+1,"\t",storage[i],"\t Rs.",price[i])

choice='y'
total=0
while (choice=='y'):
    item=int(input("Enter item number : "))
    qty=int(input("Enter quantity : "))
    total=total+price[item-1]*qty
    choice=input("Add another item? [y/n] : ")
print("Total = Rs.",total)
```
